```python

import nest_asyncio
nest_asyncio.apply()

def _get_soundcloud(r):
    for rr in r['links']:
        if 'soundcloud' in rr:
            return r['name'], rr
    return r['name'], None

class StreetParadeSoundcloudDownloader:
    """Performs the following steps:
    
    1. Parses the artists from a file, and keeps the ones that have a soundcloud (for easy scraping)
    2. Finds some links to tracks per-artist
    3. Downloads the tracks to a folder corresponding to the artist
    4. Computes embeding per-song
    5. Finds embeddings per-artist
    
    """
    artist_track_urls = {}
    def __init__(self, streetparade_html = ARTIST_HTML_FILENAME ):
        self.streetparade_html = streetparade_html
        
    def parse_artists(self):
        """
        Parse the given HTML file of artist listings and return a list of dicts
        with artist names and media links.
        """
        with open(self.streetparade_html, 'r', encoding='utf-8') as f:
            soup = BeautifulSoup(f, 'html.parser')

        artists = []
        # iterate over each artist item
        for item in soup.select('li.artist-list-item'):
            # get artist name
            name_tag = item.select_one('h3.h2')
            if not name_tag:
                continue
            name = name_tag.get_text(strip=True)

            # collect image sources
            images = []
            for img in item.select('img'):
                src = img.get('src')
                if src:
                    images.append(src)

            # collect all href links (e.g., social, site links)
            links = []
            for a in item.select('a'):
                href = a.get('href')
                if href:
                    links.append(href)

            artists.append({
                'name': name,
                'images': images,
                'links': links
            })
        self.artists = artists
        
        self.soundcloud_artists = { a[0]: a[1] for a in map(_get_soundcloud,self.artists) if a[1] is not None}
        
    def get_artist_tracks(self, artist):
        sc_url = self.soundcloud_artists[artist]
        
        asession = AsyncHTMLSession()

        async def get_coverart_links(page_url):
            r = await asession.get(page_url)
            await r.html.arender(
                sleep=15, timeout=40, scrolldown = 1,
            )
            await asyncio.sleep(5)  # belt-and-suspenders wait
            return [
                urljoin(page_url, a.attrs["href"])
                for a in r.html.find("a.sound__coverArt[href]")
            ]
        tracks_link = sc_url + '/tracks'
        print(tracks_link)
        _tracks = asyncio.run(get_coverart_links(tracks_link))
        print(f'Found {len(_tracks)} links' )
        return _tracks
        # self.artist_track_urls[artist] = await _get_track_links_per_artist(sc_url, timeout)
```
