```python
import nest_asyncio 
nest_asyncio.apply()

import sys

sys.path.append('src')
from pathlib import Path
from streetparade_embeddings.soundcloud import ArtistData, discover_track_urls_sync

artist_name = "Hilit Kolet"
artist_url = "https://soundcloud.com/hilitkolet"

cache_dir = Path(".songs_cache")
max_tracks = 5

# Discover track URLs from the artist's SoundCloud /tracks page.
track_urls = discover_track_urls_sync(artist_url)

print(f"Found {len(track_urls)} tracks")
for url in track_urls[:max_tracks]:
    print(url)

# Download and cache the first N tracks.
artist_data = ArtistData(
    artist=artist_name,
    links=track_urls,
    cache_folder=cache_dir,
)

downloads = artist_data.download_links(num_links=max_tracks)
```
